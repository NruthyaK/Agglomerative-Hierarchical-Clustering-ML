```python
from sklearn.preprocessing import StandardScaler
from sklearn.datasets import load_iris
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
iris = load_iris()
```

```python
X = iris.data
y = iris.target
```

```python
# step 1 - scale the data
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)
# X_scaled
```

```python
# step 2 - figuring the no of cluster using a dengogram
from scipy.cluster.hierarchy import linkage, dendrogram

Z = linkage(X_scaled, method="ward")
plt.figure(figsize=(10,10))
dendrogram(Z)
plt.xlabel("Samples")
plt.ylabel("Distances")
plt.title("Agglomerative Hierarchy")
```

### dendrogram is a mere indicator only. it is just a suggestive process, where we can consider using dendrogram as a way to find the total number of custer in out data

```python
# Performing Agglomerative clustering

from sklearn.cluster import AgglomerativeClustering

agg = AgglomerativeClustering(n_clusters=2)
labels = agg.fit_predict(X_scaled)

sns.scatterplot(x=X[:,0], y=X[:,2], c=labels)
```

```python
# as we know, in iris dataset, we get 3 clusters, lets plot using the actual no of clusters needed
agg = AgglomerativeClustering(n_clusters=3)
labels = agg.fit_predict(X_scaled)

sns.scatterplot(x=X[:,0], y=X[:,2], c=labels)
```
